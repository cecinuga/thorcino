```python
import numpy as np
from sklearn.datasets import make_regression

from examples.linear_regression.helpers.dataset import split_dataset

SEED = 888
BIAS = 25

IN_FEATURE, OUT_FEATURE = 15, 1

SPLIT_RATIO = 0.9

"""Signal to estimate, generated via sklearn's make_regression (noise-free, so it reduces to y = SLOPE*x + BIAS)"""
X, Y, coef = make_regression(n_samples=2, n_features=IN_FEATURE, n_targets=OUT_FEATURE, bias=BIAS, noise=0.0, coef=True, random_state=SEED)
X_tr, Y_tr = split_dataset(np.stack([X, Y]), split_ratio=SPLIT_RATIO)

def closed_form_fit(x: np.ndarray, y: np.ndarray) -> tuple[float, float]:
    """OLS closed-form solution (w, b) minimizing mse(w*x + b, y) on the given sample"""
    A = np.vstack([x, np.ones_like(x)]).T
    w, b = np.linalg.lstsq(A, y, rcond=None)[0]
    return w, b
```
